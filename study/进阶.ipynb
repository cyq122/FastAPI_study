<h1>中间件</h1>

是一个在每次请求进入FastAPI应用时都会被执行的函数，在请求到达实际路径操作之前运行，并且在响应返回给客户端之前再运行一次。<h4>作用：</h4>使用中间件为每个请求前后添加统一的处理逻辑<br>
eg：<br>
1.多个接口需要验证用户身份
2.多个接口需要记录日志、性能数据

中间件：函数顶部使用装饰器@app.middleware("http")

若存在多个中间件，则根据代码顺序自下而上执行

```python
@app.middleware("http")
async def middleware(request,call_next):
    print("中间件开始处理 -- star")
    response = await call_next(request)
    print("中间件处理完成 -- end")
    return response
```

<h1>依赖注入</h1>

抽取可复用的组件，实现代码复用、解耦且可轻松替换依赖项进行测试<br>
依赖项：可重用的组件（函数/类），负责提供某种功能或数据<br>
注入：FastAPI自动帮你调用依赖项，并将结果“注入”到路径操作函数中

1.创建依赖项<br>
2.导入depends<br>
3.声明依赖项

```python
#分页参数逻辑共用：玩具列表和工厂列表
#1.依赖项
async def common_parameters(
    skip:int=Query(default=0,ge=0),
    limit:int=Query(default=10,le=40),
):
    return {"skip":skip,"limit":limit}
#2.注入依赖
@app.get("/toys/toys_list")
async def get_toys_list(commons = Depends(common_parameters)):
    return commons

@app.get("/factory/factory_list")
async def get_factory_list(commons = Depends(common_parameters)):
    return commons
```
