<h2>路由</h2>

是URL地址和处理函数之间的映射关系，决定了当用户访问某个特定网址时，服务器应该执行哪段代码返回结果

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
async def root():
    return {"message":"hello world"}
```

<h2>参数</h2>

<h3>路径参数</h3>
位置：URL路径的一部分/book/{id}<br>
作用：指向唯一的、特定的资源<br>
方法：GET

<h3>查询参数</h3>
位置：URL？之后 k1=v1&k2=v2     http://127.0.0.1:8000/news/news_list?skip=0&limit=10<br>
作用：对资源集合进行过滤、排序、分页等操作<br>
方法：GET

<h3>请求体</h3>
位置：HTTP请求的消息体(body)中<br>
作用：创建、更新资源，携带大量数据，如：JSON<br>
方法：POST、PUT等

```python
#路径参数(URL路径的一部分)
@app.get("/book/{id}")
async def get_book1(id: int):
    return {"id":id,"title":f"这是第{id}本书"}

#类型注解Path，FastAPI允许为参数声明额外的信息和校验
"""path参数：gt/ge/lt/le——大于/大于等于/小于/小于等于
   description：描述
   min_length/max_length：长度限制
"""
from fastapi import Path
@app.get("/book/{id}")
async def get_book1(id: int = Path(...,gt=0,le=100)):
    return {"id":id,"title":f"这是第{id}本书"}

#------------------------------------------------------------
#查询参数
#查询新闻 ->分页，skip：跳过的记录数，limit：返回的记录数
@app.get("/news/news_list")
async def get_news_list(skip: int,limit:int=10):
    retuen {"skip":skip,"limit":limit}

#参数，Query函数
#------------------------------------------------------------
#请求体
#定义类型
from pydantic import BaseModel
class User(BaseModel):
    name:str
    password:int

#类型注解
@app.post("/register")
async def register(user:User):
    return user

#参数，Field函数
```

<h2>响应类型</h2>

JSONResponse--默认响应，返回JSON数据<br>
HTMLResponse--返回HTML内容<br>
FileResponse--返回文件下载

<h3>响应类型设置方式</h3>

装饰器中指定响应类<br>
场景：固定返回类型（HTML、纯文本等）

返回响应对象<br>
场景：文件下载、图片、流式响应

```python
@app.get("/html",response_class=HTMLResponse)
async def get_html():
    return "<h1>这是标题</h1>"

@app.get("/file")
async def get_file():
    file_path = "./files/1.jpeg",
    return FileResponse(file_path)
```

<h3>自定义响应数据格式</h3>


```python
from pydantic import BaseModel
class News(BaseModel):
    id:int
    title:str
    content:str

@app.get("/news/{id}",response_model=News)
async def get_news(id: int):
    return{
        "id":id,
        "title":f"这是第{id}本书",
        "content":f"这是一本Python教程"

    }
```
